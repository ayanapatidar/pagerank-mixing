## Degrees

The out-degree of a page is the number of links leaving it, and the in-degree is
the number arriving. Out-degree sets the surfer's transition probabilities (1/d).

```python
out_deg = np.asarray(A.sum(axis=1)).ravel().astype(int)
in_deg = np.asarray(A.sum(axis=0)).ravel().astype(int)

print("mean out-degree:", out_deg.mean())
print("max out-degree:", out_deg.max(), " max in-degree:", in_deg.max())
print("pages with no in-links:", (in_deg == 0).sum())
```

The mean out-degree is 8.2. Most pages have few links, and a small number have very many, which indicates two heavy tailed distributions. The largest in-degree is 38,606, and 20,315 pages (7.2%) have no in-links at all, so they can only be reached by teleports.

The largest out-degree being 255 is an artifact of a cap on links per page in the original crawl. ($2^8 -1$)
